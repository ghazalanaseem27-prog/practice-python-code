```python
# Import Pandas
import pandas as pd

# Create Customer Purchase Dataset
customer_data = {
    "Customer_ID":[101,102,103,104,105,106,107,108,109,110,111,112],

    "Customer_Name":[
        "Ali","Sara","Ahmed","Ayesha","Bilal","Fatima",
        "Usman","Hina","Zain","Noor","Hamza","Iqra"
    ],

    "City":[
        "Lahore","Karachi","Islamabad","Lahore","Karachi","Islamabad",
        "Lahore","Karachi","Islamabad","Lahore","Karachi","Islamabad"
    ],

    "Membership":[
        "Silver","Gold","Silver","Gold","Bronze","Silver",
        "Gold","Bronze","Gold","Silver","Gold","Bronze"
    ],

    "Purchase_Amount":[
        1200,2500,1800,5000,900,2200,
        4800,1100,3500,2700,2100,25000
    ]
}

# Convert to DataFrame
df = pd.DataFrame(customer_data)

# Save CSV File
df.to_csv("customer_purchase.csv", index=False)

print("CSV File Created Successfully")
```

```python
# Calculate Quartiles
Q1 = df["Purchase_Amount"].quantile(0.25)
Q3 = df["Purchase_Amount"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

print("Q1 :", Q1)
print("Q3 :", Q3)
print("IQR :", IQR)
```

```python
# Calculate First Quartile (25%)
# quantile(0.25) is used to calculate the first quartile (Q1)
Q1 = df["Purchase_Amount"].quantile(0.25)

# Calculate Third Quartile (75%)
Q3 = df["Purchase_Amount"].quantile(0.75)

# Calculate Interquartile Range
IQR = Q3 - Q1

# Calculate Lower Limit
lower_limit = Q1 - (1.5 * IQR)

# Calculate Upper Limit
upper_limit = Q3 + (1.5 * IQR)

# Display Results
print("Q1 :", Q1)
print("Q3 :", Q3)
print("IQR :", IQR)
print("Lower Limit :", lower_limit)
print("Upper Limit :", upper_limit)
```

```python
# Find Outliers
outliers = df[
    (df["Purchase_Amount"] < lower_limit) |
    (df["Purchase_Amount"] > upper_limit)
]

# Display Outliers
print(outliers)
```

```python
# skew() is used to measure the asymmetry of the data distribution
skewness = df["Purchase_Amount"].skew()
print("Skewness :", skewness)
```

```python
# kurt() is used to measure the tailedness of the data distribution
kurtosis = df["Purchase_Amount"].kurt()

# print() is used to display the output
print("Kurtosis :", kurtosis)
```

```python
import numpy as np

sales = np.array([48, 52, 49, 51, 50, 47, 53, 49, 52, 50])

sample_mean = np.mean(sales)
sample_std = np.std(sales, ddof=1)
n = len(sales)

print("Sample Mean:", sample_mean)
print("Sample Standard Deviation:", sample_std)
print("Sample Size:", n)
```

```python
from scipy import stats

confidence_level = 0.95

interval = stats.t.interval(
    confidence_level,
    df=n-1,
    loc=sample_mean,
    scale=stats.sem(sales)
)

print("95% Confidence Interval:", interval)
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error

study_hours = np.array([1, 2, 3, 4, 5, 6]).reshape(-1, 1)
marks = np.array([50, 55, 65, 70, 80, 90])

model = LinearRegression()

model.fit(study_hours, marks)

predicted_marks = model.predict(study_hours)

residuals = marks - predicted_marks

mae = mean_absolute_error(marks, predicted_marks)
mse = mean_squared_error(marks, predicted_marks)

print("Predicted Marks:", predicted_marks)
print("Residuals:", residuals)
print("MAE:", mae)
print("MSE:", mse)
```
